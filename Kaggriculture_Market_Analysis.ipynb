# 🏪 Kaggriculture 城镇商店供需机制与时序博弈深度解析

> [!NOTE] 
> **摘要 (Abstract)**
> 在 Kaggriculture 中，共享市场价格（Market Price）的崩盘是玩家最大的敌人，而**城镇商店（Unlocked Shops）**是拯救价格的唯一“排水口”。本报告详细梳理了底层需求模型，拆解了第一次开店与后续开店的深远影响，并首次引入**养殖类资产的“时间衰减”生命周期模型**。结合目前 Top 玩家 86% 同质化的 `1牛4羊` 镜像开局，揭示了从“产能比拼”向“订单微观结构博弈”演变的终极形态。


## 📊 1. 市场消耗的数学模型与基准需求底座

要理解商店的影响，首先必须掌握游戏底层的刚性通缩公式。每日的商品总需求量 `demand` 由**城镇中心（Town Center）**与**随机商店（Shops）**共同构成。

### 1.1 城镇中心：随时间膨胀的基础通缩
城镇中心作为系统兜底，持续均匀地消耗所有“非肥料”产品。随着游戏进程推进，其胃口呈阶梯式放大（常规模式下）：
- **Day 0 - Day 9 (开局/扩张期)**：每天 2 个单位。
- **Day 10 - Day 19 (复利期)**：消耗量翻倍，每天 4 个单位。
- **Day 20 - Day 29 (清算期)**：消耗量再次翻倍，每天 8 个单位。

## ⏱️ 2. 时序效应：首店指挥棒 vs. 后续补丁

商店的解锁时机对博弈环境具有决定性影响。**开局的锚定效应**与**后期的边际递减**在策略选择上截然不同。

### 2.1 第一次开店：全盘生产路线的“战略指挥棒”
第 0-4 天时，双方资金充裕、地块未定，首店直接决定了资源的全局倾斜：
* **首发单品专卖店（如 `YARN_STORE`） -> 绝对垄断路线**：羊毛有了每日 12 个的刚需，玩家可放手将羊群从 4 只扩充至 8 只以上。抛售策略可以极度放松，无需疯狂压价抢跑。
* **首发大宗混合店（如 `BAKERY`） -> 防守反击路线**：有了鸡蛋和小麦的无风险消耗池，玩家可以重仓小麦基地，以此为基础疯狂孵化鹅/鸡，稳健发育。
* **首发高危奢侈品店（如 `ICE_CREAM_SHOP`） -> 极限竞速路线**：因为草莓/牛奶极度容易线性暴跌，单靠 1 家店救不了市。玩家反而不能盲目扩产，必须激活 **2-Turn Lead（提前两回合抢跑抛售）**，宁愿工人闲置也要抢在对手前面砸盘。

### 2.2 后续开店：同质化暴击与时间衰减陷阱
* **同质化叠加 (The Boom Effect)**：因为商店是**有放回抽样**，若连续开出两家 `YARN_STORE`，全镇每日消耗 24+ 羊毛。此时羊毛变成“绝对硬通货”，无需抢跑，产多少砸多少。
* **异质化稀释 (The Dilution)**：若后续开出农贸市场（大杂烩），需求被平摊在廉价农产品上。这种店无法撑起暴利商品的抛售，只配作为多余小麦的“垃圾桶”，不应引发任何生产转型。
* **时间衰减效应 (Time Decay)**：Day 20 起，Town Center 每日自带 8 单位消耗，新店每日 6 单位的边际增量被严重稀释，不足以扭转由于全场高库存引发的暴跌风险。


## ⏳ 3. 养殖类资产在“时间衰减”下的高手动作与RoI清算

在 Kaggriculture 30 天（720 回合）的固定生命周期中，购买动物是最大的重资产投资。随着时间推移，**动物的回本期望会发生致命的“时间衰减”**。

### 3.1 牛羊的极限量化回本周期 (Break-even Lifecycle)
* **奶牛 (COW)**：成本 \$400。第 8 天首次产奶，之后每 2 天产 1 奶，基础奶价 \$160。
  * **回本要求**：需要卖出 2.5 桶奶才能回本。
  * **时间线**：第 8 天（1桶）+ 第 10 天（2桶）+ 第 12 天（3桶，回本）。**一头牛从买入到回本，需要整整 12 天。**
* **绵羊 (SHEEP)**：成本 \$500。第 6 天首次产毛，之后每 3 天产 1 毛，基础毛价 \$200。
  * **回本要求**：需要卖出 2.5 坨羊毛才能回本。
  * **时间线**：第 6 天（1坨）+ 第 9 天（2坨）+ 第 12 天（3坨，回本）。**一只羊从买入到回本，同样需要 12 天。**

### 3.2 顶尖脚本的终局清算阀值 (Closeout Horizon)
> [!WARNING]
> **新手陷阱**：在 Day 20 突然看到镇上开了一家 `PIZZA_SHOP`，兴奋地花 \$400 块买了一头牛。由于第 8 天才产首奶，这头牛在 Day 28 才产出 1 桶奶，到 Day 30 游戏强制结束。**净亏损：\$400 (本金) - \$160 (收益) = -240 金币。** 这笔投资直接将游戏拱手相让。

**高手博弈动作 (The Expert Response)**：
高级代码会在 **Day 16 ~ Day 18** 之间触发硬性熔断（Hard Stop）。
1. **停止一切重资产扩容**：绝不在 Day 18 之后买入任何牛羊或高时长种子，因为它们注定无法在 Day 30 闭市前收回成本。
2. **转向绝对提款机模式 (Harvest & Liquidate)**：将所有资金转为储备小麦（饲料），纯粹压榨现有畜群产出。
3. **彻底无视后期商店**：无论 Day 20 之后开出什么诱人的高价商店，都严厉拒绝补单转型。


## 🔥 4. 固定镜像开局下的中局分歧点与微操博弈

根据对 Top 30 顶级玩家录像的复盘，当前环境已经演化出了 86% 采用 `1 COW, 4 SHEEP, 5 WHEAT, HIRE4` 的绝对同质化镜像开局。既然前几天的操作犹如剧本般一致，真正的决定胜负的**中局分歧点（Divergence Points）**体现在以下微观层面：

### 4.1 抛售时机的极限博弈 (Queue Leading)
双方必然都在 Day 6 产出羊毛、Day 8 产出牛奶。同时抛售必然导致后者接盘暴跌（羊毛过剩 59 个砸底，牛奶过剩 76 个砸底）。
* **分歧点**：死板的脚本固定在晚上统一卖出；顶级控制流一旦发现库存就绪，不惜打断农场运作，**提前 1~2 个回合挂出卖单（2-Turn Lead）**。宁可工人闲置，也要截胡吃满 \$200 和 \$160 的溢价。单次分歧差额可达数百金币。

### 4.2 订单槽位的原位重排 (In-place Slot Ranking)
在同一个回合中，玩家通常需要同时卖出小麦（换零钱）和牛奶（换大钱）并买入饲料。
* **分歧点**：普通脚本固定了 `[SELL WHEAT, SELL MILK, BUY WHEAT]` 的顺序。而顶级脚本实现了**订单原位重排**，通过动态计算价格冲击（Impact Score），强制改写为 `[SELL MILK, SELL WHEAT, BUY WHEAT]`。
* **影响**：由于回合内订单串行结算，先卖出的牛奶不受踩踏，而后卖出的小麦抗跌能力极强（>6000承载力）。这一个小小的排序调换，在 720 回合内能累积上千硬币的利润护城河。

### 4.3 杂草的动态修复能力 (Robust Weed Repair)
农场随机生成的杂草（Weed）是死板路线的克星。
* **分歧点**：当杂草长在预定牧场格子上时，低级代理会彻底卡死死锁，畜牧断粮崩盘；顶级代理具备动态嗅探机制，自动插入 `DIG` 动作，将原计划无损顺延 1 回合。中局的容错率决定了长期胜率。


## 🔬 5. Benchmark 代理库：固定路线流的历史使命与极限预估

通过分析历代 Benchmark 代理清单和天梯对战表，犹如翻阅一部 Kaggriculture 战术演化史。其中以“元路线流”（Meta Line：统一采用 `8牛/5羊/6草莓` 的宏观规划）深刻揭示了固定路线流的作用与其未来的天花板。

### 5.1 固定路线流起到的历史作用
1. **降维打击与元博弈收敛 (Macro-planning Domination)**：
   在 720 回合内完美调度农民移动、浇水、喂养牲畜，这在计算复杂度上对传统的启发式算法或简单强化学习是灾难性的。固定路线流证明了：**一个硬编码的、被极限压榨过 720 步运筹效率的静态宏观计划，能够对纯动态代理造成碾压式的降维打击**。它们强行统一了整个高分段的生产范式，使得 86% 的玩家采用了几乎相同的开局。
2. **剥离并凸显“微操执行层”的纯粹价值 (Isolating the Execution Edge)**：
   Benchmark 库中最精妙的设计在于测试不同的执行层交易脚本。它们**拥有绝对一致的 720 步下地生产动作，最终收益却差了成千上万金币**。这充当了一个完美的控制变量实验室——证明了在产能一致的同质化时代，胜负 100% 来源于微观交易结构（卖单原位重排、抢跑时机、资金流保护）。

### 5.2 未来发展极限与天花板预估
固定路线流的红利期已经结束，继续在一套 720 步的静态指令上死磕产能，已经触及了帕累托最优的边界。未来它的发展极限与演化方向将呈现以下趋势：

1. **对“随机扰动”的极度脆弱性将成为致命伤 (The Glass Cannon Flaw)**：
   纯静态路线预设了一个绝对理想的环境。未来顶级博弈中，一旦随机刷新的杂草（Weed）长在关键格子上，或者对手通过激进的市场倾销导致你某回合少赚了 5 块钱，死板的路线就会因为“差几块钱买不起饲料”而导致动物成批饿死（Cascade Failure）。因此，**纯固定路线的天花板已经被锁死**。
2. **演化终局：静态宏观骨架 + 动态微观免疫 (Macro-skeleton + Micro-immunity)**：
   未来的天花板玩家将抛弃“纯固定”，走向“混合控制”。生产大方向依然是预先算好的最优解，但会额外包裹一层强大的“动态免疫系统”：
   * **自动拓扑修复**：遇到杂草自动插入 `DIG` 动作并在 n 回合内补齐延误，无损接轨原计划。
   * **对抗性金融战**：从“闭门造车”转向“狙击对手资金链”。预判对手的死板路线，在其急需现金买饲料的当口，恶意倾销砸穿市场底价，利用对手固定脚本无法动态变现的弱点，兵不血刃地饿死对手的畜群。
3. **基于首店的“量子态预案”取代单一预案 (Branching Contingencies)**：
   未来不再会有统一的 `1牛4羊` 唯一神阵。顶级队伍必然会在本地算力集群上预运算 3~4 套应对不同首发商店（如双毛线店特化阵、防守阵）的 720 步完美骨架，在游戏第 1 回合根据 `Unlocked Shops` 瞬间坍缩为最适合的路线，并辅以极致的市场微操执行到底。


## 🎯 6. 最终结论

在 Kaggriculture 高端局中，种地本身不再是秘密，真正的核心能力是**资本的运作与退出时机**：
1. **初期看独占，后期看堆叠**。依据首店类型决定扩容速度，绝不信任草莓/牛奶的虚假繁荣。
2. **严守 12 天回本生命线**。Day 18 之后冻结一切长线重资产投资，转入极限压榨模式。
3. **在镜像开局中重兵布置市场层**。通过 2-Turn 抢跑和槽位重排，将市场的踩踏损失全部转嫁给对手，实现零和博弈下的绝对碾压。
